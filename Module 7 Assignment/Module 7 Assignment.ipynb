## 4. Your persona and notes

```python
persona = (
    "Mod5 Bot, a rude assistant that drills you in order to make sure you understand the material."
)

reader = PdfReader("notes/notes.pdf")
notes = ""
STUDY_FILE = "study_this.txt"
NOTES_PATH = "notes/notes.pdf"
for page in reader.pages:
    text = page.extract_text()
    if text:
        notes += text

print(persona)
print(notes[:300], "...")
```

## 5. Port your first tool

**TODO:** Take your first tool from the Module 6 assignment and rewrite it as a plain, typed Python
function decorated with `@function_tool`, the same way Lab 6, Step 5a ported
`log_missed_question_tool`. Keep the function's actual behavior identical: only the decorator and
the removal of your hand written JSON schema should change.

```python
@function_tool
def my_tool(question: str) -> str:
    """Record a question the student got wrong or didn't know to study_this.txt.

    Args:
        question: The exact question the student missed.
    """
    print(f"Tool called to log when you get an answer wrong: {question}")
    with open(STUDY_FILE, "a", encoding="utf-8") as f:
        f.write(question + "\n")
    return "Wrong answer recorded"

print("Tool ready:", my_tool.name)
```

## 6. Port your second tool

**TODO:** Same idea, for your second tool (in Module 6 this was the one backed by a free, no-key
external API), the same way Lab 6, Step 5b ported `fetch_bonus_question_tool`.

```python
# TODO: your second tool, decorated with @function_tool
OPENTDB_CATEGORIES = {"general_knowledge": 9}

@function_tool
def api_tool(category: Optional[Literal["general_knowledge", "science_computers"]] = None) -> str:
    """Fetch a bonus trivia question from an external trivia database when the student asks for a break.

    Args:
        category: Trivia category to pull from (general_knowledge or science_computers).
    """
    print(f"Tool called to fetch a bonus trivia question (category={category})")
    params = {"amount": 1, "type": "multiple"}
    if category in OPENTDB_CATEGORIES:
        params["category"] = OPENTDB_CATEGORIES[category]

    try:
        response = requests.get("https://opentdb.com/api.php", params=params, timeout=20)
        data = response.json()
    except (requests.exceptions.RequestException, ValueError) as e:
        print(f"api_tool network error: {e}")
        return "The trivia API is temporarily unavailable, get back to studying your notes."

    if data.get("response_code") != 0 or not data.get("results"):
        return "Could not fetch a bonus question right now, stop slacking and get back to the quiz."

    q = data["results"][0]
    question = html.unescape(q["question"])
    answer = html.unescape(q["correct_answer"])
    return f"Bonus question: {question} (correct answer: {answer})"

print("Tool ready:", api_tool.name)
```

## 7. Port your third tool

**TODO:** Same idea, for your third, self-designed tool, the same way Lab 6, Step 5c ported
`check_progress_tool`.

```python
# TODO: your third tool, decorated with @function_tool
@function_tool
def third_tool() -> str:
    """Check how many questions have been logged as missed in study_this.txt."""
    print("Tool called to check review progress")
    try:
        with open(STUDY_FILE, "r", encoding="utf-8") as f:
            missed = [line for line in f if line.strip()]
    except FileNotFoundError:
        missed = []

    if not missed:
        return "No questions logged yet, surprisingly."
    return f"You have logged {len(missed)} missed question(s) in study_this.txt."

print("Tool ready:", third_tool.name)
```

## 8. Build your Agent

**TODO:** Build one `Agent`, the same way Lab 6, Step 6 built Study Buddy: a `name`, `instructions`
built from your persona and your notes (with the same rules your old system prompt had about when
to use each tool), a `tools` list containing all three of your ported tools, and `model=azure_model`.

Notice what is missing compared to your Module 6 assignment: no `call_tool(name, arguments)`
dispatcher, and no JSON schema for any of the three tools. `Agent` and `@function_tool` already
cover both.

```python
instructions = f"""
# Your role

You are {persona}

You ask questions on the provided notes to test knowledge.

# Context

Here are the notes you are quizzing the student on:

{notes}

# Rules

Ask exactly ONE question at a time, based only on the notes above. Wait for the student's answer
before asking the next question.

After the student answers, say clearly whether they were right or wrong, staying in character the whole time.

If the student's answer was wrong, or they say "I don't know", you must record it as a missed question using my_tool before moving on to the next question.

If the student asks for a break from the notes, use api_tool to fetch a trivia question, this does not need to be related to the notes.

If the student asks how they are doing, use third_tool to see how many questions they have logged so far.

IMPORTANT: only ask quiz questions from the notes above. If the student asks about something outside the notes that doesn't involve any other tools, tell them it's not related to the subject at hand.
"""

mod5_agent = Agent(
    name="Mod5 Bot",
    instructions=instructions,
    tools=[my_tool, api_tool, third_tool],
    model=azure_model,
)

print("Agent built:", mod5_agent.name)
```

## 10. Give it a real chat UI with Gradio

**TODO:** Adapt the `chat` function and `gr.ChatInterface` call from Lab 6, Step 8 to your own
agent: an `async def chat(message, history)` function that rebuilds `history` as a plain list of
`{"role": ..., "content": ...}` dictionaries, appends the new `message` as one more `user` turn,
passes that to `await Runner.run(your_agent, ...)`, and returns `result.final_output`, handed to
`gr.ChatInterface(chat).launch(inbrowser=True)`.

As in Lab 6, `gr.ChatInterface` accepts your async `chat` function directly, and its `history`
already comes in the shape `Runner.run` expects, so no extra conversion is needed.

Try a conversation that would trigger more than one of your three tools across a few turns, then
interrupt the cell (stop button) when you are done chatting.

```python
async def chat(message, history):
    history = [{"role": item["role"], "content": item["content"]} for item in history]
    result = await Runner.run(mod5_agent, history + [{"role": "user", "content": message}])
    return result.final_output

gr.ChatInterface(chat).launch(inbrowser=True)
```
